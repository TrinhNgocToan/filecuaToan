### **2.1. Nguồn dữ liệu**

Nguồn dữ liệu được sử dụng là bộ dữ liệu công khai **Medical Cost Personal Datasets** trên Kaggle, được lưu trữ dưới dạng file **Insurance.csv**. Nhóm tiến hành thu thập dữ liệu từ nguồn này để hình thành bộ dữ liệu thô ban đầu. Sau đó, dữ liệu được làm sạch, chuẩn hóa và chuyển đổi nhằm phục vụ cho quá trình phân tích, trực quan hóa và xây dựng mô hình dự đoán.

### **2.2. Mô tả dữ liệu**

Bộ dữ liệu gồm các biến chính như sau:

* **Age**: tuổi của người tham gia bảo hiểm.
* **Sex**: giới tính.
* **Bmi**: chỉ số khối cơ thể.
* **Children**: số con.
* **Smoker**: tình trạng hút thuốc.
* **Region**: khu vực sinh sống.
* **Charges**: chi phí bảo hiểm y tế.

Trong đó, biến **charges** là **biến đầu ra** cần dự đoán. Các biến còn lại đóng vai trò là **biến đầu vào**, được sử dụng để phân tích và xây dựng mô hình. Việc mô tả dữ liệu rõ ràng ngay từ đầu giúp xác định đúng bản chất bài toán và định hướng phù hợp cho các bước xử lý tiếp theo.

### **2.3. Làm sạch dữ liệu**

Ở bước này, nhóm tiến hành làm sạch dữ liệu để bảo đảm tính chính xác và nhất quán trước khi phân tích. Các công việc chính bao gồm:

* Kiểm tra giá trị thiếu trong toàn bộ bộ dữ liệu.
* Nếu xuất hiện dữ liệu thiếu, tiến hành xử lý bằng cách loại bỏ hoặc thay thế bằng các giá trị phù hợp như trung bình, trung vị hoặc mode tùy thuộc vào từng biến.
* Kiểm tra dữ liệu trùng lặp và loại bỏ các bản ghi trùng để tránh gây sai lệch kết quả phân tích.
* Rà soát các giá trị bất thường hoặc ngoại lệ, đặc biệt ở các biến như `bmi` và `charges`.
* Kiểm tra lại kiểu dữ liệu của từng cột để bảo đảm dữ liệu ở đúng định dạng trước khi đưa vào xử lý tiếp theo.

Bước làm sạch dữ liệu là nền móng của toàn bộ quá trình phân tích. Nếu dữ liệu đầu vào còn lỗi, mọi biểu đồ đẹp phía sau cũng chỉ là sân khấu dựng trên cát.

### **2.4. Chuẩn hóa và chuyển đổi dữ liệu**

Sau khi làm sạch, dữ liệu tiếp tục được chuẩn hóa và chuyển đổi để bảo đảm phù hợp cho trực quan hóa và xây dựng mô hình. Các nội dung chính gồm:

* Chuẩn hóa tên biến và kiểu dữ liệu về định dạng thống nhất.
* Mã hóa các biến phân loại như `sex`, `smoker`, `region` sang dạng số hoặc thực hiện **one-hot encoding** để mô hình có thể xử lý.
* Kiểm tra lại dữ liệu sau mã hóa để tránh phát sinh lỗi hoặc mất thông tin.

Bước này giúp bộ dữ liệu trở nên “nói cùng một ngôn ngữ” với các công cụ phân tích và thuật toán học máy, tránh tình trạng dữ liệu thì có đó mà mô hình lại đứng nhìn như đọc nhầm bản đồ.

### **2.5. Tạo biến mới**

Để tăng khả năng phân tích và trực quan hóa, nhóm dự kiến tạo thêm một số biến mới từ dữ liệu gốc. Cụ thể:

* **Nhóm tuổi**: chia thành các nhóm như trẻ, trung niên, cao tuổi.
* **Nhóm BMI**: chia thành thiếu cân, bình thường, thừa cân, béo phì.
* **Nhóm số con**: chia thành không có con, ít con, nhiều con.

Các biến mới này giúp hỗ trợ tốt hơn cho việc so sánh giữa các nhóm đối tượng, giúp kết quả trực quan hóa rõ ràng hơn và tạo thêm góc nhìn khi phân tích dữ liệu. Thay vì chỉ nhìn các con số rời rạc, việc phân nhóm giúp dữ liệu chuyển từ dạng hạt cát sang những cụm hình có thể kể chuyện.

### **2.6. Kết quả sau tiền xử lý**

Sau khi hoàn thành các bước tiền xử lý, nhóm kỳ vọng thu được một bộ dữ liệu:

## **4. CHỦ ĐỀ NÂNG CAO: XÂY DỰNG MÔ HÌNH DỰ ĐOÁN CHI PHÍ BẢO HIỂM Y TẾ**

### **4.1. Xác định bài toán**

Đề tài được xác định là **bài toán hồi quy**, với mục tiêu dự đoán giá trị của biến **charges** dựa trên các biến đầu vào gồm: `age`, `sex`, `bmi`, `children`, `smoker` và `region`.

Việc xác định đúng bài toán là cơ sở quan trọng để lựa chọn mô hình phù hợp, xây dựng quy trình huấn luyện chính xác và lựa chọn bộ chỉ số đánh giá hợp lý.

### **4.2. Mô hình cơ sở**

Nhóm dự kiến sử dụng **Hồi quy tuyến tính bội (Multiple Linear Regression)** làm mô hình cơ sở.

Lý do lựa chọn mô hình này gồm:

* Dễ triển khai
* Dễ giải thích
* Cho phép quan sát ảnh hưởng tuyến tính của từng biến lên `charges`
* Thích hợp để làm mô hình chuẩn so sánh với các mô hình nâng cao hơn

Mô hình cơ sở đóng vai trò như vạch xuất phát. Nó không phải cỗ máy phức tạp nhất, nhưng rất hữu ích để xem các mô hình phía sau thực sự tiến bộ đến đâu.

### **4.3. Mô hình nâng cao**

Nhóm dự kiến sử dụng **Random Forest Regression** làm mô hình nâng cao chính.

Lý do lựa chọn mô hình này gồm:

* Phù hợp với bài toán dự đoán biến số liên tục
* Xử lý tốt dữ liệu bảng có cả biến số và biến phân loại sau mã hóa
* Không giả định mối quan hệ tuyến tính giữa đầu vào và đầu ra
* Có khả năng học được các tương tác phức tạp giữa các biến, ví dụ như tương tác giữa `smoker` và `bmi`
* Ít nhạy với chuẩn hóa dữ liệu hơn so với một số mô hình khác
* Cho phép xác định mức độ quan trọng của từng biến
* Dễ triển khai trong cả R và Python

### 2.1.9 - Kiểm tra ngoại lệ bằng IQR ( những giá trị ngoại lai)

```r
def detect_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    outliers = data[(data[column] < lower_bound) | (data[column] > upper_bound)]

    print(f"\nCột: {column}")
    print(f"Q1 = {Q1}")
    print(f"Q3 = {Q3}")
    print(f"IQR = {IQR}")
    print(f"Ngưỡng dưới = {lower_bound}")
    print(f"Ngưỡng trên = {upper_bound}")
    print(f"Số lượng ngoại lệ = {outliers.shape[0]}")

    return outliers

outliers_bmi = detect_outliers_iqr(df, 'bmi')
outliers_charges = detect_outliers_iqr(df, 'charges')
```

Đoạn code sử dụng phương pháp IQR để xác định các giá trị ngoại lệ ở hai cột bmi và charges dựa trên khoảng tứ phân vị Q1 và Q3.
Kết quả cho thấy cột bmi có 9 giá trị ngoại lệ, với các giá trị nằm ngoài khoảng từ 13.675 đến 47.315.
Trong khi đó, cột charges có tới 139 giá trị ngoại lệ, với các giá trị nằm ngoài khoảng -13120.716 đến 34524.778 cho thấy chi phí bảo hiểm là biến có mức độ phân tán lớn và chứa nhiều giá trị bất thường.

### 2.1.10 - Xử lý ngoại lai

```r
def remove_outliers_iqr(data, column):
    Q1 = data[column].quantile(0.25)
    Q3 = data[column].quantile(0.75)
    IQR = Q3 - Q1

    lower_bound = Q1 - 1.5 * IQR
    upper_bound = Q3 + 1.5 * IQR

    # Giữ lại các giá trị nằm trong khoảng hợp lệ
    filtered_data = data[(data[column] >= lower_bound) & (data[column] <= upper_bound)]

    print(f"\nCột: {column}")
    print(f"Tổng số dòng : {data.shape[0]}")
    print(f"Phát hiện : {data.shape[0] - filtered_data.shape[0]} dòng có giá trị ngoại lai")

    return filtered_data


# Áp dụng
df = remove_outliers_iqr(df, 'bmi')
df = remove_outliers_iqr(df, 'charges')
```

### 2.3.1 - Tạo biến mới

```r
# Tạo nhóm tuổi
def categorize_age(age):
    if age < 30:
        return 'Thanh niên'
    elif age < 60:
        return 'Trung niên'
    else:
        return 'Cao tuổi'
df_2['age_group'] = df_2['age'].apply(categorize_age)

df_2[['age', 'age_group']].head()
```

Hàm categorize_age() được tạo ra để phân nhóm tuổi thành 3 mức: Thanh niên, Trung niên và Cao tuổi dựa trên các ngưỡng cụ thể, sau đó áp dụng hàm này lên cột age bằng .apply() để tạo biến mới age_group. Kết quả hiển thị cho thấy việc phân nhóm được thực hiện đúng logic và nhất quán với dữ liệu ban đầu

### 2.3.2 - Tạo nhóm BMI

```r
# Tạo nhóm BMI
def categorize_bmi (bmi):
    if bmi < 18.5:
        return 'Thiếu cân'
    elif bmi < 25:
        return 'Bình thường'
    elif bmi < 30:
        return 'Thừa cân'
    else:
        return 'Béo phì'

df_2['bmi_group'] = df_2['bmi'].apply(categorize_bmi)

df_2[['bmi', 'bmi_group']].head()
```

Hàm categorize_bmi được tạo ra để phân loại chỉ số BMI thành 4 nhóm: Thiếu cân, Bình thường, Thừa cân và Béo phì. Sau đó, hàm được áp dụng lên cột bmi bằng .apply() để tạo ra cột mới bmi_group. Kết quả hiển thị cho thấy việc phân loại hoạt động chính xác theo từng giá trị BMI. Đây là bước hữu ích để chuyển dữ liệu số sang dạng phân nhóm phục vụ phân tích.

### 2.3.3 - Tạo nhóm số con

```r
# Tạo nhóm số con
def categorize_children(children):
    if children == 0:
        return 'Không có con'
    elif children <= 2:
        return 'Ít con'
    else:
        return 'Nhiều con'

df_2['children_group'] = df_2['children'].apply(categorize_children)

df_2[['children', 'children_group']].head()
```

Hàm categorize_children được tạo ra để chia số lượng con thành 3 nhóm: Không có con, ít con và Nhiều con. Sau đó, dữ liệu trong cột children được xử lý bằng .apply() để sinh thêm cột children_group. Kết quả cho thấy các giá trị đã được phân loại đúng theo điều kiện đặt ra. Việc này giúp chuyển dữ liệu số sang dạng nhóm, hỗ trợ phân tích dễ dàng hơn.

### 3.3.2 - Chuẩn hóa tên biến

```r
df <- df %>%
  mutate(
    smoker = ifelse(smoker == 1, "Hút thuốc", "Không hút"),
    sex = ifelse(sex == 1, "Nữ", "Nam"),

    bmi_group = recode(bmi_group,
                       "Underweight" = "Thiếu cân",
                       "Normal" = "Bình thường",
                       "Overweight" = "Thừa cân",
                       "Obese" = "Béo phì")
  )
df %>% count(smoker)
df %>% count(sex)
df %>% count(bmi_group)
```

Sau khi chuẩn hóa dữ liệu, các biến phân loại như smoker, sex và bmi_group đã được chuyển sang dạng dễ hiểu bằng tiếng Việt. Điều này cho thấy dữ liệu đã sẵn sàng cho bước phân tích tiếp theo.

### 3.3.3 - Chuyển sang Long format

```r
df_long <- df %>%
  select(charges, smoker, sex, bmi_group) %>%
  pivot_longer(
    cols = c(smoker, sex, bmi_group),
    names_to = "variable",
    values_to = "group"
  )

head(df_long)
```

Đoạn code sử dụng pivot_longer để chuyển dữ liệu từ dạng wide sang long, giúp gom các biến smoker, sex, bmi_group vào một cột chung.
Nhờ đó, cấu trúc dữ liệu trở nên gọn gàng và linh hoạt hơn khi phân tích theo từng biến.

## 3.4 - Trực quan hóa tỷ lệ

```r
ratio_data <- insurance %>%
  mutate(
    sex = factor(sex, levels = c("Nam", "Nữ")),
    smoker = factor(smoker, levels = c("Không hút thuốc", "Hút thuốc")),
    region = factor(region, levels = c("Đông Bắc", "Tây Bắc", "Đông Nam", "Tây Nam")),
    bmi_group = factor(
      bmi_group,
      levels = c("Thiếu cân", "Bình thường", "Thừa cân", "Béo phì")
    )
  )
```

### 3.4.1 - Tỷ lệ người hút thuốc và không hút thuốc

### 3.4.4 - Tỷ lệ các nhóm BMI

```r
p_charges_hist <- ggplot(insurance, aes(x = charges)) +
  geom_histogram(bins = 30, fill = "steelblue", color = "white", linewidth = 0.3) +
  scale_x_continuous(labels = comma) +
  labs(
    title = "Histogram của charges",
    x = "Chi phí bảo hiểm",
    y = "Tần suất"
  ) +
  theme_minimal(base_size = 18)
p_charges_density <- ggplot(insurance, aes(x = charges)) +
  geom_density(fill = "tomato", color = "firebrick", alpha = 0.7, linewidth = 1) +
  scale_x_continuous(labels = comma) +
  labs(
    title = "Density Plot của charges",
    x = "Chi phí bảo hiểm",
    y = "Mật độ"
  ) +
  theme_minimal(base_size = 18)
p_charges_box <- ggplot(insurance, aes(y = charges)) +
  geom_boxplot(fill = "lightgreen", color = "darkgreen", alpha = 0.8) +
  scale_y_continuous(labels = comma) +
  labs(
    title = "Boxplot của charges",
    y = "Chi phí bảo hiểm"
  ) +
  theme_minimal(base_size = 18)
(p_charges_hist | p_charges_density) / p_charges_box
```
